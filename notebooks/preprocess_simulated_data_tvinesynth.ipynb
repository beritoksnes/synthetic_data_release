```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import json
```

```python
# Fetch dataset 
data = pd.read_csv("../data/simulated_data_tvinesynth/simulated_data_tvinesynth.csv", index_col=0)
n, d = data.shape
```

```python
data.head()
```

```python
# Define response variable
old_response = data.columns[d-1]
response = "Y"
data.rename({old_response: response}, axis=1, inplace=True)
```

```python
# Save preprocessed file 
data.to_csv("../data/simulated_data_tvinesynth/simulated_data_tvinesynth.csv", index=False)
```

```python
# Create json file
def df_to_schema(df, categorical_cols):
    columns = []
    for col in df.columns:
        if col in categorical_cols:
            categories = df[col].astype(str).unique().tolist()
            columns.append({"name": col,
                            "type": "Categorical",
                            "size": len(categories),
                            "i2s": categories})
        else:
            columns.append({"name": col,
                            "type": "Float",
                            "min": float(df[col].min()),
                            "max": float(df[col].max())})
    return {"columns": columns}

# Save json files
with open("../data/simulated_data_tvinesynth/simulated_data_tvinesynth.json", "w") as file:
    json.dump(df_to_schema(data, categorical_cols=["Y"]), file, indent=2)
```
